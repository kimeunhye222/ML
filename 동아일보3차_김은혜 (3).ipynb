```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score
```

```python
# iris data 불러오기
iris = load_iris()
```

```python
# iris data를 문제와 답으로 분리
X = iris.data 
y = iris.target
print(X,y)
```

```python
#  (3) 분리된 문제와 답을 train data와 test data로 분리
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print("전체 문제 개수:", len(X))
print("공부용 문제 개수:", len(X_train))
print("시험용 문제 개수:", len(X_test))
```

```python
 # (4) KNN분류기 모델 생성 (이웃 수는 3)
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=3)
print(knn)
```

```python
# 교차검증 실행 (5조각으로 나눠 진행)
scores = cross_val_score(knn, X, y, cv=5)
```

```python
# (6) 결과확인
print("교차 검증 점수 :", scores)
print("평균 점수 :", scores.mean())
```

```python
# 분류 문제 사례에 맞는 예시를 선택하시오
# 1번, 3번, 5번
```

```python
# 분류평가 지표 중 정확도만으로는 평가가 안 되는 경우가 있어서 정밀도와 재현
#율을 보는 경우가 있다. 항목에 맞게 사례를 선택하시오
# 정밀도가 중요한 사례(true라고 분류한 것 중에서 실제 true인 것의 비율)
# 2,4번
# 재현율이 중요한 사례(실제 true인 것 중에서 모델이 true라고 예측한 것의 비율)
# 1,4 번
```

```python
# 12-2 위 모델의 정확도, 재현율, 정밀도를 계산하시오.
TN = 95
FP = 0
FN = 5
TP = 0

# (2) 계산
accuracy = (TP + TN) / (TP + TN + FP + FN)
recall = TP / (TP + FN)

# 정밀도는 분모가 0
if (TP + FP) == 0:
    precision = "계산 불가 (분모가 0)"
else:
    precision = TP / (TP + FP)

# (3) 결과 출력
print(f"정확도: {accuracy * 100}%")
print(f"재현율: {recall * 100}%")
print(f"정밀도: {precision}")
```
